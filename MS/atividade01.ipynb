### Problema 1
*Uma caixa contém duas bolas, que podem ser ambas vermelhas, ambas
azuis, ou uma de cada cor. A cada passo, uma bola é sorteada e substituída
por uma nova bola, que tem 80% de chance de ser da mesma cor. Se
inicialmente ambas são vermelhas, qual a probabilidade de que a quinta
bola retirada seja vermelha?*

```python
import numpy as np

def calcularProbabilidadeBola():
    matrizTransicao = np.array([
        [0.8, 0.2, 0.0], 
        [0.1, 0.8, 0.1],  
        [0.0, 0.2, 0.8]   
    ])
    EstadoInicial = np.array([0, 0, 1])
    EstadoQuatro = EstadoInicial @ np.linalg.matrix_power(matrizTransicao, 4)
    # % de sucesso para cada estado
    probabilidadesSorteio = np.array([0.0, 0.5, 1.0])
    # calculo do produto escalar
    probabilidadeFinal = np.dot(EstadoQuatro, probabilidadesSorteio)

    print(f"probabilidades dos estados apos 4 passos:")
    print(f"  P(0 Vermelhas): {EstadoQuatro[0]:.4f}")
    print(f"  P(1 Vermelha):  {EstadoQuatro[1]:.4f}")
    print(f"  P(2 Vermelhas): {EstadoQuatro[2]:.4f}")
    print("-" * 40)
    print(f"probabilidade da quinta bola ser vermelha: {probabilidadeFinal * 100:.2f}%")

calcularProbabilidadeBola()
```

### Problema 2
*Numa estrada há carros e caminhões. Três de cada quatro caminhões
na estrada são seguidos por um carro, enquanto que apenas um de
cada cinco carros é seguido por um caminhão. Qual a porcentagem de
caminhões na estrada?*

```python
import random

def simular_estrada(totalVeiculos=100000):
    veiculos = []
    veiculoAtual = random.choice(['caminhão', 'carro'])
    veiculos.append(veiculoAtual)

    for _ in range(totalVeiculos - 1):
        r = random.random()
        if veiculoAtual == 'caminhão':
            # 3 de cada 4 caminhoes sao seguidos por carros
            if r < 0.75:
                veiculoAtual = 'carro'
            else:
                veiculoAtual = 'caminhão'
        else:
            #1 de cada 5 carros eh seguido por caminhao
            if r < 0.20:
                veiculoAtual = 'caminhão'
            else:
                veiculoAtual = 'carro'
        veiculos.append(veiculoAtual)
    totalCaminhao = veiculos.count('caminhão')
    totalCarro = veiculos.count('carro')
    porcentagem = (totalCaminhao / totalVeiculos) * 100

    print(f"simulação concluida com {totalVeiculos} veiculos:")
    print(f"caminhões: {totalCaminhao}")
    print(f"carros: {totalCarro}")
    print(f"porcentagem de Caminhões: {porcentagem:.2f}%")
    print(f"valor teorico esperado: {(4/19)*100:.2f}%")


simular_estrada()
```

### Problema 3
*Sabe-se que cada item produzido por uma certa máquina será defeituoso
com uma probabilidade 0,001, independentemente de qualquer outro
item. Qual a probabilidade de que em uma amostra contendo três
itens, pelo menos um esteja com defeito?*

```python
import random

def simular_probabilidade(experimentos=100000):
    sucessos = 0
    probDefeito = 0.001
    amostraTam = 3

    for _ in range(experimentos):
        amostra = [random.random() < probDefeito for _ in range(amostraTam)]
        if any(amostra):
            sucessos += 1
    probSimulada = sucessos / experimentos
    return probSimulada

resultado = simular_probabilidade()
print(f"probabilidade simulada: {resultado:.6f}")
print(f"probabilidade teorica : 0.002997")
```
