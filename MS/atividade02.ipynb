### Problema 1
*Helena está na janela de casa observando uma tempestade de raios
no horizonte. Um raio cai em média a cada 45 segundos. Se Helena
não vê nenhum raio no primeiro minuto, ela cansa de esperar e sai da
janela. Senão, ela observa por mais dois minutos.*

```python
import numpy as np

def tempestade(simulacoes):
    taxa_lambda = 4/3
    raios_total = np.random.poisson(lam = taxa_lambda,size =simulacoes)
    mask_continuam = raios_total > 0
    qtd_extra = mask_continuam.sum()
    raios_extras = np.random.poisson(lam = taxa_lambda * 2, size = qtd_extra )
    raios_total[mask_continuam] += raios_extras
    tempo_total = 1 + 2 *mask_continuam

    return raios_total.mean(), tempo_total.mean()

if __name__ == "__main__":
    n = 1000000
    est_raios, est_tempo = tempestade(n)
    print("-" * 40)
    print(f"(a) Número esperado de raios: {est_raios:.4f}")
    print("-" * 40)
    print(f"(b) Tempo esperado na janela: {est_tempo:.4f}")
    print("-" * 40)
```

###Resposta
(a) *Qual o número esperado de raios que Helena observa?*

 ≈ 3.3 raios

(b) *Qual o tempo esperado que ela fica na janela?*

 ≈ 2.5 minutos
